# In-notebook morphoview (xgraphics / xjupyter)

Requires the **xmorpho** kernel, morphoview (`xgraphics` / `xshow`), xeus-morpho’s `xjupyter` module (this repo on `~/.morphopackages`), and the jupyterlab-morpho extension (WebGL MIME renderer).

```morpho
import xgraphics
import xcolor
import xjupyter

var g = Graphics()
g.display(Sphere([0, 0, 0], 1, color=Red))
g.display(Sphere([1.5, 0, 0], 0.5, color=Color(0.2, 0.6, 1, 0.5)))
Display(g)
```
