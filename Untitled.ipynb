```morpho
// Boing — bouncing checkered ball in a wireframe room (live Graphics session).
//
//   morpho6 examples/boing.morpho
//
// Close the viewer window (or wait for the run to finish) to exit.

import morphoview
import xgraphics
import xcolor
import constants

var roomGrey = Color(10/15, 10/15, 10/15)
var roomWire = Color(1, 0, 1)                 // bright magenta grid
var ballRed = Color(1, 0, 0)
var ballWhite = Color(1, 1, 1)
var ballShadow = Color(0.22, 0.22, 0.22)

var roomHalf = 3.0
var roomTiles = 8
var wallH = 3.0
var ballR = 0.85
var wallX = roomHalf - ballR - 0.1
var nLat = 8
var nLon = 16
var floorZ = 0.2
var pitch = 35*Pi/180
var gravity = -18.0
var dt = 1/50
var nFrames = 1200

fn pitchPt(p) {
  var c = cos(pitch)
  var s = sin(pitch)
  return [p[0], p[1]*c - p[2]*s, p[1]*s + p[2]*c]
}

fn pitchN(n) {
  return pitchPt(n)
}

fn worldPos(x, y, z) {
  return pitchPt([x, y, z])
}

/* Magenta wireframe floor + back wall, pitched for default −Z camera. */
fn wireroom(half, n, height) {
  var pts = []
  var segs = []
  var h = 2*half/Float(n)
  var vh = height/Float(n)
  var zBack = -half

  // Floor: lines of constant z (span x)
  for (j in 0...(n+1)) {
    var z = -half + Float(j)*h
    var i0 = pts.count()
    pts.append(pitchPt([-half, 0, z]))
    pts.append(pitchPt([half, 0, z]))
    segs.append([i0, i0+1])
  }
  // Floor: lines of constant x (span z)
  for (i in 0...(n+1)) {
    var x = -half + Float(i)*h
    var i0 = pts.count()
    pts.append(pitchPt([x, 0, -half]))
    pts.append(pitchPt([x, 0, half]))
    segs.append([i0, i0+1])
  }
  // Wall: horizontal lines (span x)
  for (j in 0...(n+1)) {
    var y = Float(j)*vh
    var i0 = pts.count()
    pts.append(pitchPt([-half, y, zBack]))
    pts.append(pitchPt([half, y, zBack]))
    segs.append([i0, i0+1])
  }
  // Wall: vertical lines (span y)
  for (i in 0...(n+1)) {
    var x = -half + Float(i)*h
    var i0 = pts.count()
    pts.append(pitchPt([x, 0, zBack]))
    pts.append(pitchPt([x, height, zBack]))
    segs.append([i0, i0+1])
  }

  return LineSet(pts, segs, color=roomWire)
}

fn checkersphere(nlat, nlon) {
  var nv = nlat*nlon*4
  var pos = Matrix(3, nv)
  var nrm = Matrix(3, nv)
  var col = Matrix(3, nv)
  var triplets = []
  var squash = 0.86
  var vi = 0
  var fi = 0
  for (i in 0...nlat) {
    var t0 = Pi*Float(i)/Float(nlat)
    var t1 = Pi*Float(i+1)/Float(nlat)
    for (j in 0...nlon) {
      var p0 = 2*Pi*Float(j)/Float(nlon)
      var p1 = 2*Pi*Float(j+1)/Float(nlon)
      var raw = [
        [sin(t0)*cos(p0), cos(t0)*squash, sin(t0)*sin(p0)],
        [sin(t0)*cos(p1), cos(t0)*squash, sin(t0)*sin(p1)],
        [sin(t1)*cos(p1), cos(t1)*squash, sin(t1)*sin(p1)],
        [sin(t1)*cos(p0), cos(t1)*squash, sin(t1)*sin(p0)]
      ]
      var parity = i + j - 2*Int(Float(i + j)/2)
      var rgb = (parity == 0) ?
        [ballRed.r, ballRed.g, ballRed.b] :
        [ballWhite.r, ballWhite.g, ballWhite.b]
      var e1 = [raw[1][0]-raw[0][0], raw[1][1]-raw[0][1], raw[1][2]-raw[0][2]]
      var e2 = [raw[3][0]-raw[0][0], raw[3][1]-raw[0][1], raw[3][2]-raw[0][2]]
      var nx = e1[1]*e2[2]-e1[2]*e2[1]
      var ny = e1[2]*e2[0]-e1[0]*e2[2]
      var nz = e1[0]*e2[1]-e1[1]*e2[0]
      var nn = sqrt(nx*nx+ny*ny+nz*nz)
      if (nn < 1e-8) { nx=0; ny=1; nz=0; nn=1 }
      var nrmP = pitchN([nx/nn, ny/nn, nz/nn])
      for (k in 0...4) {
        var pp = pitchPt(raw[k])
        pos[0,vi]=pp[0]; pos[1,vi]=pp[1]; pos[2,vi]=pp[2]
        nrm[0,vi]=nrmP[0]; nrm[1,vi]=nrmP[1]; nrm[2,vi]=nrmP[2]
        col[0,vi]=rgb[0]; col[1,vi]=rgb[1]; col[2,vi]=rgb[2]
        vi += 1
      }
      var b = vi-4
      triplets.append([b,fi,1]); triplets.append([b+1,fi,1]); triplets.append([b+2,fi,1]); fi+=1
      triplets.append([b,fi,1]); triplets.append([b+2,fi,1]); triplets.append([b+3,fi,1]); fi+=1
    }
  }
  return TriangleComplex(pos, nrm, col, Sparse(triplets))
}

fn unitshadow() {
  var n = 28
  var pos = Matrix(3, n+1)
  var nrm = Matrix(3, n+1)
  var col = Matrix(3, n+1)
  var triplets = []
  var nn = pitchN([0, 1, 0])
  var p0 = pitchPt([0, 0.02, 0])
  pos[0,0]=p0[0]; pos[1,0]=p0[1]; pos[2,0]=p0[2]
  nrm[0,0]=nn[0]; nrm[1,0]=nn[1]; nrm[2,0]=nn[2]
  col[0,0]=ballShadow.r; col[1,0]=ballShadow.g; col[2,0]=ballShadow.b
  for (i in 0...n) {
    var a = 2*Pi*Float(i)/Float(n)
    var pp = pitchPt([cos(a), 0.02, 0.65*sin(a)])
    pos[0,i+1]=pp[0]; pos[1,i+1]=pp[1]; pos[2,i+1]=pp[2]
    nrm[0,i+1]=nn[0]; nrm[1,i+1]=nn[1]; nrm[2,i+1]=nn[2]
    col[0,i+1]=ballShadow.r; col[1,i+1]=ballShadow.g; col[2,i+1]=ballShadow.b
  }
  for (i in 0...n) {
    var j = i+1
    if (j == n) j = 0
    triplets.append([0,i,1]); triplets.append([i+1,i,1]); triplets.append([j+1,i,1])
  }
  var sh = TriangleComplex(pos, nrm, col, Sparse(triplets))
  sh.transmit = 0.35
  return sh
}

fn shadowScale(cy) {
  var height = cy - ballR
  if (height < 0) height = 0
  return ballR*(1.1 + 0.2*height)
}

var x = -1.0
var y = 2.2
var z = floorZ
var vx = 2.5
var vy = 0.0
var angle = 0.0

var g = Scene(title="Boing", background=roomGrey)

/* Translucent shadow before the room (display order matches the working
   translucent-then-vertex-color case); ball last. */
var sp = worldPos(x, 0.04, z)
var bp = worldPos(x, y, z)
var shadow = g.display(unitshadow(), sp, scale=shadowScale(y), flat=true)
g.display(wireroom(roomHalf, roomTiles, wallH))
var ball = g.display(checkersphere(nLat, nLon), bp, scale=ballR, rotate=[angle, 0, 1, 0],
                     flat=true)

var v = View(g, connecttimeout=10000)

for (frame in 1...nFrames) {
  if (v.closed) break

  vy += gravity*dt
  x += vx*dt
  y += vy*dt

  if (y < ballR) {
    y = ballR
    vy = -vy
  }
  if (x > wallX)  { x = wallX;  vx = -abs(vx) }
  if (x < -wallX) { x = -wallX; vx =  abs(vx) }

  angle += vx/ballR*dt

  g.beginBatch()
  g.move(ball, worldPos(x, y, z), rotate=[angle, 0, 1, 0])
  g.move(shadow, worldPos(x, 0.04, z), scale=shadowScale(y))
  g.endBatch()
  System.sleep(dt)
  v.poll(timeoutms=0)
}

if (!v.closed) v.close()
print "ok"
```
